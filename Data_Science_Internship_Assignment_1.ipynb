```python
# Task 1.4: Functions

import math

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True

def count_vowels(text):
    vowels = "aeiouAEIOU"
    return sum(1 for char in text if char in vowels)

def circle_area(radius):
    return math.pi * radius**2

# Testing with at least two sample inputs
print("Prime 7:", is_prime(7))
print("Prime 10:", is_prime(10))

print("Vowels in Hello:", count_vowels("Hello"))
print("Vowels in Data Science:", count_vowels("Data Science"))

print("Area of circle radius 3:", circle_area(3))
print("Area of circle radius 5:", circle_area(5))
```

**Difference Between Module, Package and Library in Python**

A module is a single Python file containing functions, classes, and variables. For example, the `math` module provides mathematical functions such as square root and factorial. A package is a collection of related Python modules organized in directories. It helps developers manage and reuse code efficiently. For example, `numpy` is commonly used for numerical computing and provides a package structure with multiple modules. A library is a collection of reusable code that helps perform specific tasks without writing everything from scratch. For example, `scikit-learn` is a machine learning library used for classification, regression, and clustering. Modules, packages, and libraries make Python programming easier, more organized, and more efficient.

```python
# Task 2: Create Orders Dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "order_id": [
        "O001", "O002", "O003", "O004",
        "O005", "O006", "O007", "O008",
        "O009", "O010", "O011", "O012"
    ],
    "customer": [
        "Aarav", "Bhavya", "Chirag", "Diya",
        "Eshan", "Farah", "Gaurav", "Hina",
        "Aarav", "Diya", "Bhavya", "Eshan"
    ],
    "product": [
        "Laptop", "Headphones", "Shoes", "Backpack",
        "Smartphone", "Watch", "Tablet", "T-shirt",
        "Mouse", "Saree", "Charger", "Sunglasses"
    ],
    "category": [
        "Electronics", "Electronics", "Fashion", "Fashion",
        "Electronics", "Fashion", "Electronics", "Fashion",
        "Electronics", "Fashion", "Electronics", "Fashion"
    ],
    "quantity": [1, 2, 1, 1, 1, 1, 1, 3, 1, 1, 2, 1],
    "price": [
        55000, 1500, 2200, 1800, 32000, 3500,
        18000, 499, 899, 4500, 650, 1200
    ],
    "order_date": [
        "2025-01-05", "2025-01-07", "2025-01-08",
        "2025-01-10", "2025-01-11", "2025-01-14",
        "2025-01-15", "2025-01-18", "2025-01-20",
        "2025-01-22", "2025-01-25", "2025-01-28"
    ]
}

df = pd.DataFrame(data)

# Save dataset as CSV
df.to_csv("orders.csv", index=False)

print("Dataset created successfully!")
print(df)
```

```python
# Display first five rows
print("First Five Rows:")
display(df.head())

# Dataset information
print("Dataset Information:")
df.info()

# Statistical summary
print("Statistical Summary:")
display(df.describe())
```

```python
df["total_amount"] = df["quantity"] * df["price"]

print(df[["order_id", "customer", "quantity",
          "price", "total_amount"]])
```

```python
total_revenue = df["total_amount"].sum()

print("Total Revenue: ₹", total_revenue)
```

```python
category_revenue = df.groupby("category")["total_amount"].sum()

print("Revenue by Category:")
print(category_revenue)
```

```python
customer_spending = (
    df.groupby("customer")["total_amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 3 Customers by Total Spending:")
print(customer_spending.head(3))
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])

print(df[["order_id", "order_date"]])
print("\nData Type of order_date:")
print(df["order_date"].dtype)
```

```python
prices = np.array(df["price"])

print("All Prices:", prices)
print("Mean Price:", np.mean(prices))
print("Maximum Price:", np.max(prices))
print("Minimum Price:", np.min(prices))
print("Standard Deviation:", np.std(prices))
```

```python
# Task 3.1: EDA Summary

print("Number of rows and columns:", df.shape)
print("Number of unique customers:", df["customer"].nunique())
print("Number of unique products:", df["product"].nunique())

print("\nOrders per category:")
print(df["category"].value_counts())

print("\nMost purchased category:")
print(df["category"].value_counts().idxmax())

print("\nHighest revenue category:")
print(df.groupby("category")["total_amount"].sum().idxmax())

print("\nAverage Order Value:", df["total_amount"].mean())

print("\nAny missing values?")
print("Yes" if df.isnull().any().any() else "No")
```

```python
# Task 3.2.1: Bar Chart

category_revenue = df.groupby("category")["total_amount"].sum()

plt.figure(figsize=(8, 5))
plt.bar(category_revenue.index, category_revenue.values,
        color=["skyblue", "orange"])

plt.title("Total Revenue per Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue (INR)")
plt.tight_layout()
plt.show()
```

```python
# Task 3.2.2: Pie Chart

category_orders = df["category"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    category_orders.values,
    labels=category_orders.index,
    autopct="%1.1f%%",
    colors=["lightblue", "lightcoral"],
    startangle=90
)

plt.title("Share of Orders by Category")
plt.tight_layout()
plt.show()
```

```python
# Task 3.2.3: Count Plot

plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="category", hue="category", palette="Set2", legend=False)

plt.title("Count of Orders per Category")
plt.xlabel("Category")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

```python
# Task 3.2.4: Box Plot

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="category", y="price", hue="category", palette="Set2", legend=False)

plt.title("Price Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Price (INR)")
plt.tight_layout()
plt.show()
```

```python
# Task 3.2.5: Histogram

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="total_amount", bins=8, color="purple", kde=True)

plt.title("Distribution of Total Order Amount")
plt.xlabel("Total Amount (INR)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```
